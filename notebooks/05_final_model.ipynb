# Medical Abstracts TC Corpus — final model

The selected configuration uses TF-IDF with word unigrams and bigrams,
`min_df=2`, and Logistic Regression with balanced class weights and C=10.

These settings were selected using validation macro F1.

The model is refitted on the combined training and validation datasets.
The reserved test set is used only for final evaluation and will not
guide further hyperparameter selection.

Results apply to the filtered single-label dataset.

```python
from pathlib import Path

import joblib
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, f1_score
from sklearn.pipeline import Pipeline

RANDOM_STATE = 42
TEXT_COL = "medical_abstract"
TARGET_COL = "condition_label"

CURRENT_DIR = Path.cwd()
PROJECT_ROOT = (
    CURRENT_DIR.parent
    if CURRENT_DIR.name == "notebooks"
    else CURRENT_DIR
)

PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"
MODELS_DIR = PROJECT_ROOT / "models"
```

## 1. Loading the final training and test datasets

Training and validation data are combined for the final fit.
The test set remains separate, including during TF-IDF fitting.

```python
train_df = pd.read_csv(PROCESSED_DATA_DIR / "train.csv")
validation_df = pd.read_csv(PROCESSED_DATA_DIR / "validation.csv")
test_df = pd.read_csv(PROCESSED_DATA_DIR / "test.csv")
labels_df = pd.read_csv(PROCESSED_DATA_DIR / "labels.csv")

final_train_df = pd.concat(
    [train_df, validation_df],
    ignore_index=True,
)

expected_labels = set(labels_df[TARGET_COL])

for name, dataset in {
    "final_train": final_train_df,
    "test": test_df,
}.items():
    assert {TEXT_COL, TARGET_COL}.issubset(dataset.columns)
    assert dataset[[TEXT_COL, TARGET_COL]].notna().all().all()
    assert dataset[TEXT_COL].str.strip().ne("").all()
    assert not dataset[TEXT_COL].duplicated().any()
    assert set(dataset[TARGET_COL]) == expected_labels

    print(f"{name}: {dataset.shape}")

assert set(final_train_df[TEXT_COL]).isdisjoint(
    test_df[TEXT_COL]
)
```

## 2. Training the selected configuration

The full pipeline is fitted on the combined training and validation
data. TF-IDF learns a new vocabulary from this combined dataset,
without using test documents.

```python
final_model = Pipeline(
    steps=[
        (
            "tfidf",
            TfidfVectorizer(
                ngram_range=(1, 2),
                min_df=2,
            ),
        ),
        (
            "classifier",
            LogisticRegression(
                C=10.0,
                class_weight="balanced",
                max_iter=1000,
                random_state=RANDOM_STATE,
            ),
        ),
    ]
)

final_model.fit(
    final_train_df[TEXT_COL],
    final_train_df[TARGET_COL],
)
```

## 3. Final test evaluation

Test metrics estimate performance on the reserved split.
They are reported without using them to revise the selected configuration.

```python
test_predictions = final_model.predict(test_df[TEXT_COL])

test_metrics = {
    "accuracy": accuracy_score(
        test_df[TARGET_COL],
        test_predictions,
    ),
    "macro_f1": f1_score(
        test_df[TARGET_COL],
        test_predictions,
        average="macro",
        zero_division=0,
    ),
}

display(pd.Series(test_metrics, name="test"))

label_mapping = labels_df.set_index(TARGET_COL)[
    "condition_name"
].to_dict()

label_order = sorted(label_mapping)

test_report = pd.DataFrame(
    classification_report(
        test_df[TARGET_COL],
        test_predictions,
        labels=label_order,
        target_names=[label_mapping[label] for label in label_order],
        output_dict=True,
        zero_division=0,
    )
).T

display(test_report)
```

## 4. Saving and verifying the model

The fitted pipeline is saved together with its label mapping and test
metrics. Reloading the artifact must preserve its predictions.

Only load joblib artifacts from trusted sources.

```python
MODELS_DIR.mkdir(parents=True, exist_ok=True)
MODEL_PATH = MODELS_DIR / "medical_triage_model.joblib"

artifact = {
    "pipeline": final_model,
    "label_mapping": label_mapping,
    "test_metrics": test_metrics,
}

joblib.dump(artifact, MODEL_PATH)

loaded_artifact = joblib.load(MODEL_PATH)
reloaded_predictions = loaded_artifact["pipeline"].predict(
    test_df[TEXT_COL]
)

assert (reloaded_predictions == test_predictions).all()

print("Model saved:", MODEL_PATH)
print("Reloaded predictions match the original predictions.")
```

## 5. Conclusions

The final model was trained on 7,053 examples using the configuration
selected during validation: TF-IDF with unigrams and bigrams, min_df=2,
and Logistic Regression with balanced class weights and C=10.

On the reserved test set of 1,245 examples, the model achieved:

- Accuracy: 0.8129
- Macro F1: 0.8084

Test macro F1 was approximately 0.52 percentage points below the
previous validation result. The results are broadly consistent,
although the final model was fitted on a larger training dataset.

Neoplasms and cardiovascular diseases achieved the highest F1-scores.
General pathological conditions remained the most challenging class.

The fitted pipeline, label mapping, and test metrics were exported.
Reloading the artifact preserved all test predictions.

These results apply to the filtered single-label subset. The model
classifies medical categories, not clinical urgency.

The next stage is to serve the saved pipeline through a FastAPI
application and measure inference latency.

```python
import json
import platform

import sklearn

metadata = {
    "model_file": MODEL_PATH.name,
    "python_version": platform.python_version(),
    "sklearn_version": sklearn.__version__,
    "pandas_version": pd.__version__,
    "joblib_version": joblib.__version__,
    "training_samples": len(final_train_df),
    "test_samples": len(test_df),
    "random_state": RANDOM_STATE,
    "parameters": {
        "ngram_range": [1, 2],
        "min_df": 2,
        "C": 10.0,
        "class_weight": "balanced",
        "max_iter": 1000,
    },
    "test_metrics": test_metrics,
}

METADATA_PATH = MODELS_DIR / "model_metadata.json"

METADATA_PATH.write_text(
    json.dumps(metadata, indent=2, ensure_ascii=False),
    encoding="utf-8",
)

print("Metadata saved:", METADATA_PATH)
```
